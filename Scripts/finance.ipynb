```python
q1=df['TransactionAmount'].quantile(0.25)
q3=df['TransactionAmount'].quantile(0.75)
```

```python
q1
```

```python
q3
```

```python
#InterQuatileRange
```

```python
IQR=round(q3-q1)
```

```python
IQR
```

```python
upper_limit= q3+ 0.7*IQR
lower_limit= q1- 0.7*IQR
```

```python
upper_limit
```

```python
lower_limit
```

```python
#store Outlier
```

```python
outlier=df[(df['TransactionAmount']>upper_limit) | (df['TransactionAmount']<lower_limit)].count()
outlier
```

```python
#drop outliers
```

```python
df=df[(df['TransactionAmount']>lower_limit)&(df['TransactionAmount']<upper_limit)]
```

```python
df.count()
```
